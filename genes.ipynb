```python
import introns
```

```python
#genome_file = '/home/halakuc/genomy/longa_rascaf.fasta'
#genes_file = '/home/halakuc/longa/important/longa_stringtie_strand_informed.gtf'
genome_file = 'bugtest.fasta'
genes_file = 'bugtest.gtf'
longa_genome = introns.read_genome(genome_file)
```

```python
longa_genes = introns.read_genes(genes_file, 'stringtie')
for gene in longa_genes.values():
    gene.add_exons()
    gene.extract_sequence(longa_genome)
    print(gene.name)
```

```python
for gene in longa_genes.values():
    gene.create_introns()
```

```python
exon1 = introns.Exon('scaffold_5', 100, 105)
exon2 = introns.Exon('scaffold_5', 110, 115)
exon3 = introns.Exon('scaffold_5', 120, 125)
intron12 = introns.Intron('scaffold_5', 106, 109)
intron23 = introns.Intron('scaffold_5', 116, 119)
exon1.next_exon = exon2
exon2.next_exon = exon3
exon3.prev_exon = exon2
exon2.prev_exon = exon1
intron12.prev_exon = exon1
intron23.prev_exon = exon2
intron12.next_exon = exon2
intron23.next_exon = exon3
exon1.next_intron = intron12
exon2.prev_intron = intron12
exon2.next_intron = intron23
exon3.prev_intron = intron23


gene = introns.Gene('scaffold_5', 100, 125, strand='+', exons=[exon1, exon2, exon3])
```

```python
gene.extract_sequence(longa_genome)
gene.create_introns()
```
