## Długości parowania na pozycjach +2, +1, +3

```python
def pairing_score_for_classifier(sequence, return_sequence = False):
    def pairing_length(s, start=0, end=-1):
        n = len(s[start:end])/2
        x, y = start, end
        pairing_length = 0
        while x<n:
            if introns.complimentary(s[x], s[y]):
                pairing_length+=1
                x += 1
                y -= 1
            else:
                break
        paired_sequence = (s[start:x], s[y+1:end+1][::-1])
        return pairing_length, paired_sequence


    p2, seq2 = pairing_length(sequence, 3, -6)
    p1, seq1 = pairing_length(sequence, 4, -6)
    p3, seq3 = pairing_length(sequence, 2, -6)
    if return_sequence: return [seq2, seq1, seq3]
    return p2, p1, p3

pXs = []
for i in introns_NK_demulti:
    pXs.append(pairing_score_for_classifier(i.sequence))
p2s, p1s, p3s = [x[0] for x in pXs], [x[1] for x in pXs], [x[2] for x in pXs]
    
```

```python
def return_pairings_freqs(seqs):
    def count_GT_pairings(seq):
        pairs = zip(seq[0], seq[1])
        gt, gc, at = 0, 0, 0
        for x, y in pairs:
            if x+y == "GT" or x+y == "TG":
                gt += 1
            elif x+y == "GC" or x+y == "CG":
                gc += 1
            elif x+y == "AT" or x+y == "TA":
                at += 1
            else:
                print(x, y)
        return gt, gc, at, len(seq[0])

    gts, gcs, ats, ns = 0, 0, 0, 0
    for s in seqs:
        gt, gc, at, n = count_GT_pairings(s)
        gts += gt
        gcs += gc
        ats += at
        ns += n
    return gts, gcs, ats, ns
```

## Zależność liczby/długości parowań z przesunięciem o 2 od przesunięcia o 1 i o 3

```python
def intron_pairing_score(sequence, whether_weighted_scores = False):
    def pairing_length(s, start=0, end=-1):
        n = len(s[start:end])/2
        x, y = start, end
        best_pairing_length = 0.
        pairing_length = 0.
        total_pairings = 0.
        while x<=20:
            if total_pairings > 20: print("x =", x, total_pairings)
            if introns.complimentary(s[x], s[y]):
                #print(s[x], s[y], introns.weigh_pairings(s[x], s[y]))
                pairing_length += 1. if whether_weighted_scores==False else introns.weigh_pairings(s[x], s[y])
                total_pairings += 1. if whether_weighted_scores==False else introns.weigh_pairings(s[x], s[y])
            else: #if the pair is not complementary
                best_pairing_length = max(best_pairing_length, pairing_length)
                pairing_length = 0.
            #przesuniecie do nastepnej pary
            x += 1
            y -= 1
        best_pairing_length = max(best_pairing_length, pairing_length)
        if whether_weighted_scores == True:
            best_pairing_length, total_pairings = best_pairing_length, total_pairings
        return best_pairing_length, total_pairings
    
    pl2, tp2 = pairing_length(sequence, 0, -3) #przesuniecie o 2
    pl1, tp1 = pairing_length(sequence, 0, -2) #przesuniecie o 1
    pl3, tp3 = pairing_length(sequence, 0, -4) #przesuniecie o 3
    return [pl2, pl1, pl3, tp2, tp1, tp3]

def get_introns_list_scores(sequences, whether_weighted_scores = False):
    return pd.DataFrame(np.array([([i]+intron_pairing_score(i.sequence, whether_weighted_scores = whether_weighted_scores)) for i in sequences]),
                        columns=["intron_object", "pl2", "pl1", "pl3", "tp2", "tp1", "tp3"])

pairings_all_nk_introns = get_introns_list_scores(introns_NK_demulti)
weighed_pairings_all_nk_introns = get_introns_list_scores(introns_NK_demulti, whether_weighted_scores = True)

pairings_conventional_introns = get_introns_list_scores(introns_K_demulti)
weighed_pairings_conventional_introns = get_introns_list_scores(introns_K_demulti, whether_weighted_scores = True)
```

```python
def make_heatmap_from_pls_tps(pls_tps):
    pls_tps_df = pd.DataFrame(0., index = np.arange(21), columns = np.arange(21))
    # df z zerami, za każdą parę (x=parowanie w 1/3, y=parowanie w 2) dodajemy 1 do df'a
    for _, (x, y) in pls_tps.iterrows():
        #print(x,y)
        pls_tps_df.iloc[int(x),int(y)] += 1
    sns.heatmap(pls_tps_df, annot=True)
```

## 48 versions of classifiers

```python
wariantyNK = []
for intron in introns_K_demulti + introns_NK_demulti:
    if intron.is_man_K():
        wariantyNK.append(intron)
    wariantyNK += [var for var in intron.variations if not var.is_man_NK()]
```

```python
def compute_one_prediction_variant(features_variant, choosing_function_type, whether_weighted_scores, class_0_type):
    '''features_variant = [1-6], choosing_function_type = [pl, tp],
    whether_weighted_scores = [True, False], class_0_type = [NKvar, K]'''
    def compute_experimental_intron_characteristics(seq, features_variant,
                                                    choosing_function_type,
                                                    whether_weighted_scores):
        if type(seq)==introns.Intron:
            prev_exon_seq = seq.prev_exon.sequence[-5:]
            intron_seq = seq.sequence
            next_exon_seq = seq.next_exon.sequence[:5]
            i = seq
        elif type(seq)==str:
            prev_exon_seq = seq[:5]
            intron_seq = seq[5:-5]
            next_exon_seq = seq[-5:]
        baseY = {'C', 'T'}
        baseR = {'A', 'G'}

        if features_variant == 1: e_y_cnt, i_r_cnt, i_CAG_cnt, i_y_cnt, e_r_cnt = 0, 0, 0, 0, 0
        elif features_variant == 2: e_y_cnt, i_r_cnt, i_CAG_cnt, i_y_cnt, e_r_cnt, x2 = 0, 0, 0, 0, 0, 0
        elif features_variant == 3: e_y_cnt, i_r_cnt, i_CAG_cnt, i_y_cnt, e_r_cnt, x2, x1 = 0, 0, 0, 0, 0, 0, 0
        elif features_variant == 4: e_y_cnt, i_r_cnt, i_CAG_cnt, i_y_cnt, e_r_cnt, x2, x3 = 0, 0, 0, 0, 0, 0, 0
        elif features_variant == 5: e_y_cnt, i_r_cnt, i_CAG_cnt, i_y_cnt, e_r_cnt, x2, x1, x3 = 0, 0, 0, 0, 0, 0, 0, 0
        elif features_variant == 6: e_y_cnt, i_r_cnt, i_CAG_cnt, i_y_cnt, e_r_cnt, choosing_function = 0, 0, 0, 0, 0, 0
        
        if len(prev_exon_seq)<1: return False
        if prev_exon_seq[-1] in baseY: e_y_cnt = 1
        if intron_seq[0] in baseR: i_r_cnt = 1
        if intron_seq[3:6]=="CAG" and intron_seq[-8:-5]=="CTG": i_CAG_cnt = 1
        if intron_seq[-1] in baseY: i_y_cnt = 1
        if next_exon_seq[0] in baseR: e_r_cnt = 1
        pl2, pl1, pl3, tp2, tp1, tp3 = intron_pairing_score(seq, whether_weighted_scores = whether_weighted_scores)
        if choosing_function_type == "pl":
            x2, x1, x3 = pl2, pl1, pl3,
            choosing_function = max(pl2, pl1*0.6, pl3*0.6)
        elif choosing_function_type == "tp":
            x2, x1, x3 = tp2, tp1, tp3
            choosing_function = max(tp2, tp1*0.6, tp3*0.6)
            
        if features_variant == 1: return np.array([e_y_cnt, i_r_cnt, i_CAG_cnt, i_y_cnt, e_r_cnt])
        elif features_variant == 2: return np.array([e_y_cnt, i_r_cnt, i_CAG_cnt, i_y_cnt, e_r_cnt, x2])
        elif features_variant == 3: return np.array([e_y_cnt, i_r_cnt, i_CAG_cnt, i_y_cnt, e_r_cnt, x2, x1])
        elif features_variant == 4: return np.array([e_y_cnt, i_r_cnt, i_CAG_cnt, i_y_cnt, e_r_cnt, x2, x3])
        elif features_variant == 5: return np.array([e_y_cnt, i_r_cnt, i_CAG_cnt, i_y_cnt, e_r_cnt, x2, x1, x3])
        elif features_variant == 6: return np.array([e_y_cnt, i_r_cnt, i_CAG_cnt, i_y_cnt, e_r_cnt, choosing_function])
        
    introns_variants_characteristics, true_classes = [], []
    # introny NK - 1
    # introny I i inne pozycje LUB introny K - 0

    for intron_obj in introns_NK_demulti:
        introns_variants_characteristics.append(
            compute_experimental_intron_characteristics(intron_obj.sequence,
                                                        features_variant,
                                                        choosing_function_type,
                                                        whether_weighted_scores)
            )
        true_classes.append(1)

    if class_0_type == "NKvar": set_0 = wariantyNK
    elif class_0_type == "K": set_0 = introns_K_demulti
    for intron_obj in set_0:
        introns_variants_characteristics.append(
            compute_experimental_intron_characteristics(intron_obj.sequence,
                                                        features_variant,
                                                        choosing_function_type,
                                                        whether_weighted_scores)
            )
        true_classes.append(0)

    if features_variant == 1: columns = ['eY| i', 'e |Ri', 'CAG - CGT', 'iY| e', 'i |Re'] 
    elif features_variant == 2: columns = ['eY| i', 'e |Ri', 'CAG - CGT', 'iY| e', 'i |Re', 'x2']
    elif features_variant == 3: columns = ['eY| i', 'e |Ri', 'CAG - CGT', 'iY| e', 'i |Re', 'x2', 'x1']
    elif features_variant == 4: columns = ['eY| i', 'e |Ri', 'CAG - CGT', 'iY| e', 'i |Re', 'x2', 'x3']
    elif features_variant == 5: columns = ['eY| i', 'e |Ri', 'CAG - CGT', 'iY| e', 'i |Re', 'x2', 'x1', 'x3']
    elif features_variant == 6: columns = ['eY| i', 'e |Ri', 'CAG - CGT', 'iY| e', 'i |Re', 'pairing_function']
            
    introns_characteristics = pd.DataFrame(introns_variants_characteristics, columns = columns)
    X_train, X_test, y_train, y_test = train_test_split(introns_characteristics, true_classes,
                                                        test_size=0.3, random_state=420)
    
    print("Feature variant:", features_variant,
          "Choosing function type:", choosing_function_type,
          "weighted scores:", whether_weighted_scores,
          "0 class type:", class_0_type)
    
    clf = GaussianProcessClassifier(1.0 * RBF(1.0))
    clf.fit(X_train, y_train)
    y_test_predicted = clf.predict(X_test)
    acc = balanced_accuracy_score(y_test, y_test_predicted)
    
    return acc, clf
```

```python
typki = []
for features_variant in range(1,7):
    for choosing_function_type in ["pl", "tp"]:
        for whether_weighted_scores in [True, False]:
            for class_0_type in ["NKvar", "K"]:
                acc, clf = compute_one_prediction_variant(features_variant, choosing_function_type,
                                                     whether_weighted_scores, class_0_type)
                typki.append([features_variant, choosing_function_type,
                              whether_weighted_scores, class_0_type, acc, clf])
        
print(len(typki))
```
