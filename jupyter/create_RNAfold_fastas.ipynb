```python
import introns
import numpy as np
import matplotlib.pyplot as plt
from Bio import motifs
from Bio.Seq import Seq
from Bio import SeqIO
import re
```

```python
'''Robienie plikow z polaczonymi koncami intronow'''
path="/home/semik/projekty/noncanonical_introns/introny_joinedends/"

listaplikow = [("all", "./introny_fasty/gatunki_introny_all.fasta"), ("onesided","./introny_fasty/gatunki_introny_onesided.fasta"), ("over0", "./introny_fasty/gatunki_introny_over0.fasta"), ("unmoved-onesided","./introny_fasty/gatunki_introny_unmoved-onesided.fasta")][-1:]
for typlisty, plikfasta in listaplikow:
    print(typlisty)
    seqs=[]
    for record in SeqIO.parse(plikfasta, "fasta"):
        seq_poczatek=str(record.seq[:35])
        seq_koniec = str(record.seq[-35:])
        seq=seq_poczatek+"AAAAAAAAAA"+seq_koniec
        if typlisty=="unmoved-onesided": nazwagat, q, klasa, klasaCon = record.description.split()
        else: nazwagat, q, klasa = record.description.split()
        seqs.append((nazwagat, klasa, seq))
    for gatunek in ['E.gracilis', 'E.hiemalis', 'E.longa']:
        inst1=[seq[2] for seq in seqs if seq[0]==gatunek]
        
        print(typlisty, gatunek)
        #with open(path+'%s_%s_joinedends.fasta' %(typlisty, gatunek), 'w') as file:
        with open(path+'%s_%s_joinedends_35.fasta' %(typlisty, gatunek), 'w') as file:
            for seq in inst1:
                file.write(">intron\n"+seq+"\n")
        file.close()
```

```python
print("\nLOGA Z PODZIALEM NA KLASY\n")

'''loga z podzialem na klasy'''
listaplikow = [("all", "./introny_fasty/gatunki_introny_all.fasta")]
for typlisty, plikfasta in listaplikow:
    joinedends=[]
    for record in SeqIO.parse(plikfasta, "fasta"):
        seq_poczatek=str(record.seq[:25])
        seq_koniec = str(record.seq[-25:])
        seq=seq_poczatek+"AAAAAAAAAA"+seq_koniec
        nazwagat, q, klasa = record.description.split()
        joinedends.append((nazwagat, klasa, seq))
    for gatunek in ['E.gracilis', 'E.hiemalis', 'E.longa']:
        print(gatunek)
        seqs=[]
        for nrklasy in range(1,10):
            inst1=[seq[2] for seq in joinedends if seq[0:2]==(gatunek, str(nrklasy))]
            seqs+=inst1
            
            print(gatunek, nrklasy)
            with open(path+"klasy_%s_klasa%d_joinedends.fasta" %(gatunek, nrklasy), 'w') as file:
                for seq in inst1:
                    file.write(">intron\n"+seq+"\n")
            file.close()
        
        print(gatunek, "razem")
        with open(path+"klasy_%s_razem_joinedends.fasta" %gatunek, 'w') as file:
            for seq in seqs:
                file.write(">intron\n"+seq+"\n")
        file.close()
```
